# Experiment notebook

Calls the configured gateway through the `wotan.gateway_client` SDK.

```python
import asyncio
from wotan.gateway_client import chat, extract_json

async def main():
    answer = await chat('Say pong and nothing else.')
    print(answer)
    data = await extract_json('Maria Silva works at Acme Corp', schema={
        'type': 'object',
        'properties': {'people': {'type': 'array', 'items': {'type': 'string'}}},
    })
    print(data)

asyncio.run(main())
```
